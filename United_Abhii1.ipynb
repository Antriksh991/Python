1. Creating and Printing Lists

A list is used to store many items in one variable. We use square brackets [] to make a list, and we separate items with commas.

```python
numbers = [10, 20, 30, 40, 50]
print(numbers)
```

2.List Index(Accessing Elements)

Each items in a list has a position number called an index. Python starts counting positions from 0.



Lists: [10, 20, 30, 40, 50,]

Index:  0    1   2   3   4

Example:

```python
marks = [90, 85, 78]
print(marks[0])
```

3. Negative Indexing

Negative indexing means starting from the end of the list. -1 refers to the last item, -2 refers to the second-last item, and so on.

List:     [10, 20, 30, 40, 50]

Negative Index: -5 -4 -3 -2 -1

Example:

```python
names = ["Amit", "Priya", "Vijay"]
print(names[-1])
```

4. Changing List Elements

You can change the value of a specific element by using its index and assigning a new value.


Example:

```python
colors = ["Red", "Blue", "Green"]
colors[1] = "Yellow"
print(colors)
```

5. Adding Elements with append()


The append() method adds an element to the very end of an existing list.


Example:

```python
animals = ["cat", "dog"]
animals.append("cow")
print(animals)
```

6. Adding Elements with insert()

The insert() methods adds an element at a specific index. Its takes two arguments:insert(index, value).





Example:

```python
numbers = [10, 30, 40]
numbers.insert(1, 20)
print(numbers)
```

7. Removing Elements with remove()

The remove() method removes the first occurrence of a specific value from the list.

Example:

```python
names = ["Amit", "Rahul", "Amit"]
names.remove("Amit")
print(names)
```

8. Removing Elements with pop()

. list.pop() removes and returns the last elements of the list.

.list.pop(index) removes the elements at that specific index.

Example:

```python
letters = ["A", "B", "C"]
letters.pop()
print(letters)
```

9.Finding Length with len() and checking with in

. len(list) returns the total number of items in a list.

. value in list returns True if the item is present, otherwise False.

Example:

```python
items = ["Books", "Pens"]
print(len(items))
print("Pens" in items)
```

10. Joining (+) and Repeating (*) Lists

. Use the + operator to merge or join two lists together.

. Use the * operator to repeat a list a certain number of items.

Example:

```python
a = [1, 2]
b = [3, 4]
print(a + b)
print(a * 3)
```

11. Slicing a List

Slicing allows you to get a part of a list using list[start:stop].

Note that the stop index is not included.

Example:

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
```

12. sort(), reverse(), and clear()

.sort() arranges a list in ascending order.

.reverse() reverse the order of the elements in-place.

.clear() deletes all elements from the list, leaving it empty.

Example:

```python
vals = [3, 1, 2]
vals.sort()
print(vals)
```

13. for Loop with Lists

You can easily look at every item in a list one by one using a standard for loop.



Example:

```python
fruits = ["Apple", "Banana"]
for fruits in fruits:
  print(fruits)
```
